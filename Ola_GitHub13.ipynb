```python
with open("pride.txt", "r", encoding="utf-8") as f:
    text = f.read().lower()

print(text[:1000])
```

# **2.Build a Simple Text Generation Model**


```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
import numpy as np

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
encoded = tokenizer.texts_to_sequences([text])[0]

vocab_size = len(tokenizer.word_index) + 1
print("Vocab size:", vocab_size)
```

```python
#Create sequences

seq_length = 10
sequences = []

for i in range(seq_length, len(encoded)):
    seq = encoded[i-seq_length:i+1]
    sequences.append(seq)

sequences = np.array(sequences)
X, y = sequences[:, :-1], sequences[:, -1]
y = to_categorical(y, num_classes=vocab_size)
```

```python
#Build model

model = tf.keras.Sequential([
    tf.keras.layers.Embedding(vocab_size, 64, input_length=seq_length),
    tf.keras.layers.LSTM(128),
    tf.keras.layers.Dense(vocab_size, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
```

# **3- Text Generation Function**

```python
def generate_text(seed_text, num_words):
    result = seed_text
    for _ in range(num_words):
        tokens = tokenizer.texts_to_sequences([result])[0]
        tokens = tokens[-seq_length:]
        padded = tf.keras.preprocessing.sequence.pad_sequences([tokens], maxlen=seq_length)
        pred = model.predict(padded, verbose=0)
        next_word = tokenizer.index_word[np.argmax(pred)]
        result += " " + next_word
    return result

print(generate_text("she walked into the room", 30))
```

# **4- Application Demonstration**

```python
seed = "in the quiet evening"
print(generate_text(seed, 40))
```
